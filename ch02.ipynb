    \end{split}
\end{equation*}
$$

$$
\begin{equation*}

$$

$$
\begin{equation*}
    \begin{split}

          &= \frac{69}{i}.
    \end{split}
\end{equation*}
$$

Now, using the better approximation $\ln{(1 + r)} \approx r - \frac{1}{2}r^2$, we have for $r \approx .08$ we conclude that

$$
\begin{equation*}
    \begin{split}
        n &\approx \frac{\ln{2}}{r - \frac{1}{2}r^2} \\
          &= \frac{.69}{.08 - \frac{1}{2}{.08}^2}.

\end{equation*}
$$

Using the following calculation

```python
n = .69 / (.08 - (.5 * (.08**2)))
print(n)
```

## Exercise 4.

We have the following definition of Newton's method and the appropriate functions:

```python
from typing import Callable


def newtons_method(f: Callable[[float], float], f_p: Callable[[float], float], lambda_k: float) -> float:
    return lambda_k - (f(lambda_k) / f_p(lambda_k))


lambdas: list[float] = [1.0]


def f(x: float) -> float:
    return -1.0 + x + x**2


def f_p(x: float) -> float:
    return 1.0 + 2.0 * x


for k in range(1, 5):
    lambdas.append(newtons_method(f, f_p, lambdas[k - 1]))


for lambda_bar in lambdas:
    print(f"lambda_bar = {lambda_bar}, f(lambda_bar) = {f(lambda_bar)}")
```

Thus we conclude that the approximation after four estimates is $\bar{\lambda} \approx 0.618033988749989$ and $f(\bar{\lambda}) \approx 2.1066481892262345 \times {10}^{-13} \approx 0$.

$$

$$

## Exercise 8

To calculate the IRR on an incremental basis, we subtract each cash flow of the less expensive of the two options considered from the more expensive option, yielding positive cash flows for all but the initial payment. Thus we have the following table of cash flows:

| |Option A vs Option B|Option B vs Option C|Option A vs Option C|
|-|--------|--------|--------|
|$x_0$|-24,000|-5,000|-29,000|
|$x_1 \cdots x_4$|6,000|400|6,400|
|$x_5$|10,000|2,000|12,000|

Thus we have the following polynomial for considering changing from Option A to Option B:

$$
    -24,000 + 6,000c + 6,000c^2 + 6,000c^3 + 6,000c^4 + 10,000c^5,
$$

or, equivalently


$$
    p(r) = -24,000 + \frac{6,000}{1 + r} + \frac{6,000}{{(1 + r)}^2} + \frac{6,000}{{(1 + r)}^3}
    + \frac{6,000}{{(1 + r)}^4} + \frac{10,000}{{(1 + r)}^5}.
$$

Using Newton's method from Exercise 4, we have the following calculation:

```python
def cash_flows(r: float) -> float:
    return (
        -24000
        + 6000 / (1 + r)
        + 6000 / (1 + r) ** 2
        + 6000 / (1 + r) ** 3
        + 6000 / (1 + r) ** 4
        + 10000 / (1 + r) ** 5
    )


def cash_flows_p(r: float) -> float:
    return (
        -6000 / (1 + r) ** 2
        - 12000 / (1 + r) ** 3
        - 18000 / (1 + r) ** 4
        - 24000 / (1 + r) ** 5
        - 50000 / (1 + r) ** 6
    )

rs = [0.1]
for k in range(1, 6):
    rs.append(newtons_method(cash_flows, cash_flows_p, rs[k - 1]))

for r in rs:
    print(f"r = {r}, f(r) = {cash_flows(r)}")
```

Thus we have $r_0 \approx 0.118$, i.e., an IRR of approximately 11.8%, indicating that a change from A to B is justified on the basis of the IRR.

$$

```python
pv = 1000 * (1 - (1 / 1.05)**5) * 5
print(pv)
```
