## Exercise 2.

We have 

$$
\begin{equation*}
    \begin{split}
                             {(1 + r)}^n &= 2 \\
        \Leftrightarrow \ln{{(1 + r)}^n} &= \ln{2} \\ 
           \Leftrightarrow n\ln{(1 + r)} &= \ln{2} \\
                       \Leftrightarrow n &= \frac{\ln{2}}{\ln{(1 + r)}}.
    \end{split}
\end{equation*}
$$

Then, applying the approximation $ln{(1 + r)} \approx r$ for small $r$, we have

$$
\begin{equation*}
    \begin{split}
        n &\approx \frac{\ln{2}}{r} \\
          &= \frac{.69}{r}.
    \end{split}
\end{equation*}
$$

Thus, using $i = 100r$, we have

$$
\begin{equation*}
    \begin{split}
        n &\approx \frac{\ln{2}}{r} \\
          &= \frac{.69}{(i / 100)} \\
          &= \frac{69}{i}.
    \end{split}
\end{equation*}
$$

Now, using the better approximation $\ln{(1 + r)} \approx r - \frac{1}{2}r^2$, we have for $r \approx .08$ we conclude that

$$
\begin{equation*}
    \begin{split}
        n &\approx \frac{\ln{2}}{r - \frac{1}{2}r^2} \\
          &= \frac{.69}{.08 - \frac{1}{2}{.08}^2}.
    \end{split}
\end{equation*}
$$

Using the following calculation

```python
n = .69 / (.08 - (.5 * (.08**2)))
print(n)
```

## Exercise 4.

We have the following definition of Newton's method and the appropriate functions:

```python
from typing import Callable


def newtons_method(f: Callable[[float], float], f_p: Callable[[float], float], lambda_k: float) -> float:
    return lambda_k - (f(lambda_k) / f_p(lambda_k))


lambdas: list[float] = [1.0]


def f(x: float) -> float:
    return -1.0 + x + x**2


def f_p(x: float) -> float:
    return 1.0 + 2.0 * x


for k in range(1, 5):
    lambdas.append(newtons_method(f, f_p, lambdas[k - 1]))


for lambda_bar in lambdas:
    print(f"lambda_bar = {lambda_bar}, f(lambda_bar) = {f(lambda_bar)}")
```

Thus we conclude that the approximation after four estimates is $\bar{\lambda} \approx 0.618033988749989$ and $f(\bar{\lambda}) \approx 2.1066481892262345 \times {10}^{-13} \approx 0$.

## Exercise 5.

Given payments $x_k = 500000$, for $k = 0, 1, \ldots, 20$, and $r = 0.1$, we have the following present value: 

$$
\begin{equation*}
    \begin{split}
        \text{PV} &= \sum_{k = 0}^n{\frac{x_k}{{(1 + r)}^k}} \\
                  &= \sum_{k = 0}^{20}{\frac{500000}{1.1^k}} \\
                  &\approx 4756781.86,
    \end{split}
\end{equation*}
$$

so that the present value of the prize is $4,756,781.86.

We use the following to calculate the approximate present value:

```python
n = 20
r = 0.1
pv = 0.0
x_k = 500000
for k in range(0, n + 1):
    pv += x_k / ((1 + r) ** k)

print(pv)
```

## Exercise 6.

Assuming that the couple wants to stay in the apartment for 6 months, we have two cash flow streams $\mathbf{x} = (1000, 1000, 1000, 1000, 1000, 1000)$ and $\mathbf{z} = (1000, 900, 900, 900, 900, 900)$, with $r = 0.12$, and $k = 0, 1, \ldots, 6 = n$, so that we have the following present values:

$$
\begin{equation*}
    \begin{split}
        {\text{PV}}_{\mathbf{x}} &= \sum_{k = 0}^n{\frac{x_k}{{(1 + r)}^k}} \\
                                 &= \sum_{k = 0}^6{\frac{1000}{1.12^k}} \\
                                 &\approx 5111.41,
    \end{split}
\end{equation*}
$$

and

$$
\begin{equation*}
    \begin{split}
        {\text{PV}}_{\mathbf{z}} &= \sum_{k = 0}^n{\frac{z_k}{{(1 + r)}^k}} \\
                                 &= 1000 + \sum_{k = 1}^6{\frac{900}{1.12^k}} \\
                                 &\approx 4700.27,
    \end{split}
\end{equation*}
$$

so that the couple should choose to move to the cheaper apartment.

We use the following to calculate the approximate present values:

```python
n = 6
r = 0.12

pv_x = 0.0
xs = [1000] * (n + 1)
for k in range(0, n + 1):
    pv_x += xs[k] / ((1 + r) ** k)

print(pv_x)

pv_z = 0.0
zs = [1000] + ([900] * (n + 1))
for k in range(0, n + 1):
    pv_z += zs[k] / ((1 + r) ** k)

print(pv_z)
```
