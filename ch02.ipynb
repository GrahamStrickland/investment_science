## Exercise 1.

**(a)** Suppose we invested $1 in 1776 at 3.3% interest compounded yearly and we assume the current date to be 2009 (the date of publication), then we have for $r = 0.033$ and $n = (2009 - 1) - (1776 + 1) = 231$ years. This calculation allows for the yearly interest at the start of 2009 and accounts for the fact that interest only begins to accrue from 1777. Thus we have the present value of the investment given by

$$
\begin{equation*}
    \begin{split}
        PV &= {(1 + r)}^n \\
           &= {1.033}^{231} \\
           &\approx 1807.90.
    \end{split}
\end{equation*}
$$

We use the following to calculate the approximate value:

```python
pv = 1.033**231
print(pv)
```

The closest estimate is thus $1,000.

$$

    \begin{split}
                             {(1 + r)}^n &= 2 \\

           \Leftrightarrow n\ln{(1 + r)} &= \ln{2} \\
                       \Leftrightarrow n &= \frac{\ln{2}}{\ln{(1 + r)}}.
    \end{split}
\end{equation*}
$$

Then, applying the approximation $ln{(1 + r)} \approx r$ for small $r$, we have

$$
\begin{equation*}
    \begin{split}
        n &\approx \frac{\ln{2}}{r} \\

    \end{split}
\end{equation*}
$$
